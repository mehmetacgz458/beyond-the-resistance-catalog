```python
import pandas as pd

df = pd.read_csv("../data/raw/kpneumoniae_ast_2026-11-22.tsv", sep="\t")

print(df.shape)
df.head()
```

```python
df.info()
```

```python
from collections import Counter

sayac = Counter()
for hucre in df["AST phenotypes"]:
    for parca in str(hucre).split(","):
        if "=" in parca:
            ilac, sonuc = parca.split("=", 1)
            sayac[(ilac.strip(), sonuc.strip())] += 1

for ilac in ["meropenem", "ciprofloxacin"]:
    print(ilac, {s: n for (i, s), n in sayac.items() if i == ilac})
```

```python
kume = df["SNP cluster"].value_counts()

print("Kümesi olan izolat :", df["SNP cluster"].notna().sum())
print("Kümesi olmayan     :", df["SNP cluster"].isna().sum())
print("Farklı küme sayısı :", len(kume))
print("Tek izolatlık küme :", (kume == 1).sum())
print("2+ izolatlık küme  :", (kume >= 2).sum())
print("\nEn kalabalık 10 küme:")
print(kume.head(10))
```
